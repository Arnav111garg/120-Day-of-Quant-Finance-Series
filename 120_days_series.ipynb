##Properties Of Normal Distribution
*   **Symmetric distribution** - The normal distribution is symmetric about its
mean point. It means the distribution is perfectly balanced toward its mean point with half of the data on either side.
*   **Bell-Shaped curve** - The graph of a normal distribution takes the form bell-shaped curve with most of the points accumulated at its mean position. The shape of this curve is determined by the mean and standard deviation of the distribution.
*   **Empirical Rule** - The normal distribution curve follows the empirical rule where 68% of the data lies within 1 standard deviation from the mean of the graph, 95% of the data lies within 2 standard deviations from the mean and 99.7% of the data lies within 3 standard deviations from the mean.
*   **Additive Rule** - The sum of two or more normal distributions will always be a normal distribution.
*   **Central Limit Theoram** - It states if we take the mean of large no data points collected from independent and identical distributed random variables then these mean will follow a normal distribution regardless of their original distribution.

# MONTE CARLO SIMULATION

Monte Carlo Simulation is a computational technique that uses random sampling to estimate numerical results. The method is named after the famous Monte Carlo Casino in Monaco due to its use of randomness and chance.

At its core, Monte Carlo simulation helps model uncertainty and solve problems that might be deterministic in theory but difficult to compute directly.

**Key Applications:**
- Option pricing (e.g. European & Exotic Options under Black-Scholes or stochastic volatility models)
- Risk Management (e.g. Value at Risk (VaR))
- Portfolio Optimization under uncertain returns
- Asset price modeling using Geometric Brownian Motion (GBM)
- Credit risk and default probability simulation
- Estimating expected shortfall or tail risk

**Basic Example – Estimating Pi:**
Let’s say you want to estimate the value of π using Monte Carlo simulation:
- Imagine a unit square with a quarter circle inside.
- Randomly generate points in the square.
- Count how many fall inside the circle.
- Ratio of circle points to total ≈ π/4 → Multiply by 4 to estimate π.

```python
import numpy as np
import matplotlib.pyplot as plt

def monte_carlo_pi(num_samples):
    x = np.random.uniform(-1, 1, num_samples)
    y = np.random.uniform(-1, 1, num_samples)

    distance = np.sqrt(x**2 + y**2)

    inside_circle = np.sum(distance <= 1)

    pi_estimate = (inside_circle / num_samples) * 4

    return pi_estimate

num_samples = 10000

pi_estimate = monte_carlo_pi(num_samples)

print(f"Estimated value of π with {num_samples} samples: {pi_estimate}")
x = np.random.uniform(-1, 1, num_samples)
y = np.random.uniform(-1, 1, num_samples)
plt.figure(figsize=(6, 6))
plt.scatter(x, y, c=np.sqrt(x**2 + y**2) <= 1, cmap='bwr', s=1)
plt.gca().set_aspect('equal', adjustable='box')
plt.title('Monte Carlo Simulation for π')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## Conditional Probabiltiy

Conditional probability refers to the probability of an event A occurring given that another event B has already occurred.

It’s denoted as:

P(A | B) = Probability of A given B
Formula:

P(A | B) = P(A ∩ B) / P(B), provided P(B) > 0

**Interpretation:**

We narrow our "universe" to situations where B has happened, and then we ask: how often does A happen in that restricted universe?

## Bayes' Theorem

Bayes’ Theorem helps us reverse conditional probabilities. It tells how to update our beliefs about an event after seeing new evidence.

Formula:
P(A | B) = [P(B | A) × P(A)] / P(B)

Where:
- P(A) = prior probability (before seeing data)
- P(B | A) = likelihood (probability of observing B given A is true)
- P(B) = total probability of B (evidence)
P(A | B) = posterior probability (updated belief after seeing B)


```python
import numpy as np

# Number of simulations
n = 1_000_000

# Simulate dice rolls
die1 = np.random.randint(1, 7, size=n)
die2 = np.random.randint(1, 7, size=n)

# Define events
A = (die1 + die2 >= 10)
B = (die1 >= 4)

# Conditional probability
P_A_given_B = A[B].mean()

# Theoretical comparison
P_B = B.mean()
P_A_and_B = (A & B).mean()
P_A_given_B_theoretical = P_A_and_B / P_B

print("Simulated P(A|B):", P_A_given_B)
print("Theoretical P(A|B):", P_A_given_B_theoretical)
```

```python
P_Disease = 0.01
P_No_Disease = 0.99

P_Pos_given_Disease = 0.99
P_Pos_given_No_Disease = 0.05
P_Pos = (P_Pos_given_Disease * P_Disease) + (P_Pos_given_No_Disease * P_No_Disease)
P_Disease_given_Pos = (P_Pos_given_Disease * P_Disease) / P_Pos
print(f"P(Disease | Positive Test): {P_Disease_given_Pos:.3f}")
import matplotlib.pyplot as plt
labels = ['Disease', 'No Disease']
posterior = [P_Disease_given_Pos, 1 - P_Disease_given_Pos]
plt.figure(figsize=(2,2))
plt.bar(labels, posterior, color=['red', 'green'])
plt.title("Probability after Positive Test (Bayes’ Theorem)")
plt.ylabel("Probability")
for i, v in enumerate(posterior):
    plt.text(i, v + 0.01, f"{v:.2%}", ha='center')
plt.ylim(0, 1)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Parameters
P0 = 100
mu = 0.05        # annual drift
sigma = 0.2      # annual volatility
T = 1            # years
n_steps = 252
dt = T / n_steps

# Simulate returns
returns = np.random.normal(
    loc=mu * dt,
    scale=sigma * np.sqrt(dt),
    size=n_steps
)

# Price path
price = P0 * np.exp(np.cumsum(returns))

plt.plot(price)
plt.xlabel("Time steps")
plt.ylabel("Price")
plt.title("Asset Price Simulation under Normal Returns")
plt.show()

```

```python
n = 1_000_000

daily_returns = np.random.normal(
    loc=mu / 252,
    scale=sigma / np.sqrt(252),
    size=n
)

confidence_level = 0.05  # 5% VaR

VaR_5 = -np.quantile(daily_returns, confidence_level)

print("5% Daily VaR:", VaR_5)
```

```python
from scipy.stats import norm

z_alpha = norm.ppf(confidence_level)

VaR_parametric = -(mu/252 + sigma/np.sqrt(252) * z_alpha)

print("Parametric VaR (Normal):", VaR_parametric)
```

#A/B TESTING

"A/B testing" is a shorthand for a simple randomized controlled experiment, in which a number of samples (e.g. A and B) of a single vector-variable are compared. A/B tests are widely considered the simplest form of controlled experiment, especially when they only involve two variants. However, by adding more variants to the test, its complexity grows.

# COMMON TEST STATISTICS

Two-sample hypothesis tests are appropriate for comparing the two samples in which the samples are divided by the two control cases in the experiment. Z-tests are appropriate for comparing means under stringent conditions regarding normality and a known standard deviation. Student's t-tests are appropriate for comparing means under relaxed conditions when less is assumed. Welch's t-test assumes the least and is therefore the most commonly used two-sample hypothesis test in which the mean of a metric is to be optimized. While the mean of the variable to be optimized is the most common choice of estimator, others are regularly used.

```python
import numpy as np
from scipy import stats

np.random.seed(42)

n_days = 252

# Strategy A (baseline)
returns_A = np.random.normal(
    loc=0.0005,   # mean daily return
    scale=0.01,
    size=n_days
)

# Strategy B (slightly improved)
returns_B = np.random.normal(
    loc=0.0008,
    scale=0.01,
    size=n_days
)
```

```python
mean_A = returns_A.mean()
mean_B = returns_B.mean()

print("Mean return A:", mean_A)
print("Mean return B:", mean_B)
print("Difference:", mean_B - mean_A)
```

```python
t_stat, p_value = stats.ttest_ind(
    returns_B,
    returns_A,
    equal_var=False  # Welch's t-test
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
p_value_one_sided = p_value / 2 if t_stat > 0 else 1.0

print("One-sided p-value:", p_value_one_sided)
```

```python
alpha = 0.05

if p_value_one_sided < alpha:
    print("Reject H0: Strategy B outperforms Strategy A")
else:
    print("Fail to reject H0: No significant outperformance")
```

```python
effect_size = mean_B - mean_A
annualized_effect = effect_size * 252

print("Daily effect:", effect_size)
print("Annualized effect:", annualized_effect)
```

```python
import numpy as np
import pandas as pd

n_days = 252
strategy_returns = np.random.normal(loc=0.0005, scale=0.01, size=n_days)
returns = pd.Series(strategy_returns)
n = len(returns)
train = returns[:int(0.6 * n)]
test = returns[int(0.6 * n):int(0.8 * n)]
holdout = returns[int(0.8 * n):]

def performance_metrics(returns):
    mean = returns.mean()
    vol = returns.std()
    sharpe = mean / vol * np.sqrt(252)

    equity = (1 + returns).cumprod()
    running_max = equity.cummax()
    drawdown = (equity - running_max) / running_max

    return {
        "mean": mean,
        "vol": vol,
        "sharpe": sharpe,
        "max_drawdown": drawdown.min()}

metrics_train = performance_metrics(train)
metrics_test = performance_metrics(test)
metrics_holdout = performance_metrics(holdout)
```

```python
from scipy import stats

t_stat, p_value = stats.ttest_1samp(test, 0)

print("Test set p-value:", p_value)
```

```python
mean = test.mean()
std = test.std()
n = len(test)

ci_low = mean - 1.96 * std / np.sqrt(n)
ci_high = mean + 1.96 * std / np.sqrt(n)

print("95% CI for mean return:", (ci_low, ci_high))
```

```python
def bootstrap_mean_ci(returns, n_boot=5000):
    boot_means = [
        np.mean(np.random.choice(returns, size=len(returns), replace=True))
        for _ in range(n_boot)
    ]
    return np.percentile(boot_means, [2.5, 97.5])

bootstrap_ci = bootstrap_mean_ci(test.values)
print("Bootstrap CI:", bootstrap_ci)
```

```python
# Compute Mean Return Vector
mean_returns = returns.mean(axis=0)
print(mean_returns)
```

```python
# Center the Data
demeaned_returns = returns - mean_returns
demeaned_returns.shape
```

```python
# Covariance Matrix
cov_matrix = (
    demeaned_returns.T @ demeaned_returns) / (T - 1)
cov_matrix.shape
```

```python
# Sanity Checks
# Must be True.
# If not → your implementation is wrong.

print(np.allclose(cov_matrix, cov_matrix.T))
```

```python
variances = np.diag(cov_matrix)
print(variances)
```

```python
eigenvalues = np.linalg.eigvals(cov_matrix)
print(eigenvalues)
```

```python
np_cov = np.cov(returns, rowvar=False)

print(np.allclose(cov_matrix, np_cov))
```

```python
weights = np.array([0.25, 0.25, 0.25, 0.25])

portfolio_variance = weights.T @ cov_matrix @ weights
portfolio_vol = np.sqrt(portfolio_variance)

print("Portfolio variance:", portfolio_variance)
print("Portfolio volatility:", portfolio_vol)
```

```python
mean = returns.mean(axis=0)
std = returns.std(axis=0)

X = (returns - mean) / std
cov_matrix = (X.T @ X) / (T - 1)
cov_matrix.shape
```

```python
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
explained_variance_ratio = eigenvalues / eigenvalues.sum()

print("Explained variance ratio:")
print(explained_variance_ratio)
cumulative_variance = np.cumsum(explained_variance_ratio)
print("Cumulative variance:")
print(cumulative_variance)
```

#💡DAY 21(PROJECT): "Factor-Based Portfolio Optimizer"🚀

```python
import numpy as np
#Simulating Asset Returns
T = 500
n_assets = 6

returns = np.random.normal(
    loc=0.0005,
    scale=0.01,
    size=(T, n_assets))
```

```python
# Building the Covariance Matrix
returns_demeaned = returns - returns.mean(axis=0)
Sigma = (returns_demeaned.T @ returns_demeaned) / (T - 1)
```

```python
# Extracting Statistical Factors via PCA
eigenvalues, eigenvectors = np.linalg.eigh(Sigma)

idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Building the Factor Model
k = 2  # number of factors
B = eigenvectors[:, :k]
Sigma_f = np.diag(eigenvalues[:k])
```

```python
'''Idiosyncratic Risk(Idiosyncratic risk, also known as unsystematic risk or firm-specific risk,
 refers to the risk inherent to a specific company, industry, or asset, rather than the overall market).'''

Sigma_eps = Sigma - B @ Sigma_f @ B.T
```

```python
# Portfolio Risk Function
def portfolio_variance(w, Sigma):
    return w.T @ Sigma @ w
```

```python
# Factor Exposure Function
def factor_exposure(w, B):
    return B.T @ w
```

```python
# Optimization Setup
from scipy.optimize import minimize

n = n_assets
w0 = np.ones(n) / n

# Constraints and Bounds
constraints = ({
    'type': 'eq',
    'fun': lambda w: np.sum(w) - 1
})
bounds = [(0, 1) for _ in range(n)]

# Running the Optimization
result = minimize(
    portfolio_variance,
    w0,
    args=(Sigma,),
    constraints=constraints,
    bounds=bounds)
w_opt = result.x
print(factor_exposure(w_opt, B))
```

```python
# Risk Decomposition
factor_risk = w_opt.T @ B @ Sigma_f @ B.T @ w_opt
idiosyncratic_risk = w_opt.T @ Sigma_eps @ w_opt
```

# What the Black–Scholes Model Is:

***What is the fair price today of a European option, assuming no arbitrage?***

# Key ideas behind it:

* Prices evolve randomly but continuously

* Risk can be dynamically hedged

* In a no-arbitrage world, option prices are risk-neutral expectations

* Volatility, not expected return, drives option value


# ***Black–Scholes assumes:***

* Underlying follows Geometric Brownian Motion

* Constant volatility
𝜎
σ

* Constant risk-free rate
𝑟
r

* No arbitrage

* No dividends (for now)

* European exercise only

# The Black–Scholes Formula

***Parameters***

* So : current stock price

* 𝐾: strike price

* 𝑇: time to maturity (in years)

* 𝑟: risk-free interest rate

* σ: volatility

```python
import numpy as np
from scipy.stats import norm

# BLACK-SCHOLES CALL FUNCTION
def black_scholes_call(S0, K, T, r, sigma):
    d1 = (np.log(S0 / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)

    call_price = S0 * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
    return call_price
```

```python
# Black–Scholes Put Function
def black_scholes_put(S0, K, T, r, sigma):
    d1 = (np.log(S0 / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)

    put_price = K * np.exp(-r * T) * norm.cdf(-d2) - S0 * norm.cdf(-d1)
    return put_price
```

```python
# EXAMPLE
S0 = 100      # current price
K = 100       # strike
T = 1.0       # 1 year
r = 0.05      # 5% risk-free rate
sigma = 0.2   # 20% volatility

call = black_scholes_call(S0, K, T, r, sigma)
put = black_scholes_put(S0, K, T, r, sigma)

print("Call Price:", call)
print("Put Price:", put)
```

# ***What Implied Volatility Really Is***


From Black–Scholes, we learned:

Option Price = BS(So, K, T, r, σ)


**In markets:**

* 𝑆o , 𝐾, 𝑇, 𝑟 are observable

* Option price is observable

* 𝜎 is not

**👉 Implied volatility is the value of
𝜎 that makes the Black–Scholes price equal to the market price.**

# Why Implied Volatility Matters

* Options are quoted in IV, not price

* Volatility surfaces encode market beliefs

* Smiles and skews reveal risk asymmetry

* Almost all volatility trading is IV-based

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.optimize import brentq

# Black–Scholes Call Price
def black_scholes_call(S0, K, T, r, sigma):
    #Black–Scholes price for a European call option.
    if sigma <= 0 or T <= 0:
        return 0.0

    d1 = (np.log(S0 / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)

    return S0 * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
```

```python
# Implied Volatility Solver
def implied_volatility_call(C_market, S0, K, T, r):
#Computes implied volatility by inverting Black–Scholes.

    def objective(sigma):
        return black_scholes_call(S0, K, T, r, sigma) - C_market

    # Volatility bounds: [near zero, very high]
    return brentq(objective, 1e-6, 3.0)
```

```python
# Example
S0 = 100.0
r = 0.05
T = 1.0
true_vol = 0.25

strikes = np.linspace(70, 130, 15)
market_prices = [
    black_scholes_call(S0, K, T, r, true_vol)
    for K in strikes
]
```

```python
# Compute Implied Volatility Curve
implied_vols = np.array([
    implied_volatility_call(C, S0, K, T, r)
    for C, K in zip(market_prices, strikes)
])
implied_vols = np.clip(implied_vols, 1e-6, None)
```

```python
plt.figure(figsize=(4,4))
plt.plot(strikes, implied_vols, marker='o')
plt.axhline(true_vol, linestyle='--', label="True Volatility")
plt.xlabel("Strike")
plt.ylabel("Implied Volatility")
plt.title("Implied Volatility vs Strike (Black–Scholes World)")
plt.legend()
plt.show()
```

# CONCLUSION

Since option prices were generated using a constant-volatility Black–Scholes model, the extracted implied volatility curve is flat across strikes.

This confirms internal consistency of the pricing and inversion pipeline. Any deviations from flatness would indicate either numerical error or model misspecification.

#💡Day 31 (Code):  "**Compute Greeks numerically and compare with Black–Scholes closed-form values**"🚀

```python
import numpy as np
from scipy.stats import norm

S = 100.0      # Underlying price
K = 100.0      # Strike
T = 1.0        # Time to maturity (years)
r = 0.05       # Risk-free rate
sigma = 0.2    # Volatility
```

```python
def black_scholes_call(S, K, T, r, sigma):
    if T <= 0 or sigma <= 0:
        return 0.0

    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)

    return S * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
```

```python
# DELTA
def delta_bs(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    return norm.cdf(d1)

# GAMMA
def gamma_bs(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    return norm.pdf(d1) / (S * sigma * np.sqrt(T))

# THETA
def theta_bs(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    term1 = - (S * norm.pdf(d1) * sigma) / (2 * np.sqrt(T))
    term2 = - r * K * np.exp(-r * T) * norm.cdf(d2)
    return term1 + term2

# VEGA
def vega_bs(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    return S * np.sqrt(T) * norm.pdf(d1)

# RHO
def rho_bs(S, K, T, r, sigma):
    d2 = (np.log(S / K) + (r - 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    return K * T * np.exp(-r * T) * norm.cdf(d2)
```

```python
h_S = 1e-4      # Price step
h_sigma = 1e-4 # Volatility step
h_T = 1e-4     # Time step
h_r = 1e-4     # Rate step
```

```python
def delta_fd(S, K, T, r, sigma):
    return (black_scholes_call(S + h_S, K, T, r, sigma) -
            black_scholes_call(S - h_S, K, T, r, sigma)) / (2 * h_S)

def gamma_fd(S, K, T, r, sigma):
    return (black_scholes_call(S + h_S, K, T, r, sigma)
            - 2 * black_scholes_call(S, K, T, r, sigma)
            + black_scholes_call(S - h_S, K, T, r, sigma)) / (h_S ** 2)

def theta_fd(S, K, T, r, sigma):
    return (black_scholes_call(S, K, T - h_T, r, sigma)
            - black_scholes_call(S, K, T + h_T, r, sigma)) / (2 * h_T)

def vega_fd(S, K, T, r, sigma):
    return (black_scholes_call(S, K, T, r, sigma + h_sigma)
            - black_scholes_call(S, K, T, r, sigma - h_sigma)) / (2 * h_sigma)

def rho_fd(S, K, T, r, sigma):
    return (black_scholes_call(S, K, T, r + h_r, sigma)
            - black_scholes_call(S, K, T, r - h_r, sigma)) / (2 * h_r)
```

```python
print("Greek     Analytical      Numerical")
print("-------------------------------------")

print(f"Delta   {delta_bs(S,K,T,r,sigma):.6f}    {delta_fd(S,K,T,r,sigma):.6f}")
print(f"Gamma   {gamma_bs(S,K,T,r,sigma):.6f}    {gamma_fd(S,K,T,r,sigma):.6f}")
print(f"Theta   {theta_bs(S,K,T,r,sigma):.6f}    {theta_fd(S,K,T,r,sigma):.6f}")
print(f"Vega    {vega_bs(S,K,T,r,sigma):.6f}    {vega_fd(S,K,T,r,sigma):.6f}")
print(f"Rho     {rho_bs(S,K,T,r,sigma):.6f}    {rho_fd(S,K,T,r,sigma):.6f}")
```

#💡Day 32 (Code): "**Option Greeks Calculator**"🚀

```python
import numpy as np
from scipy.stats import norm

def d1(S, K, T, r, sigma):
    return (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))

def d2(S, K, T, r, sigma):
    return d1(S, K, T, r, sigma) - sigma * np.sqrt(T)
```

```python
def call_price(S, K, T, r, sigma):
    return S * norm.cdf(d1(S, K, T, r, sigma)) \
           - K * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma))

def put_price(S, K, T, r, sigma):
    return K * np.exp(-r * T) * norm.cdf(-d2(S, K, T, r, sigma)) \
           - S * norm.cdf(-d1(S, K, T, r, sigma))
```

```python
def delta_call(S, K, T, r, sigma):
    return norm.cdf(d1(S, K, T, r, sigma))

def delta_put(S, K, T, r, sigma):
    return norm.cdf(d1(S, K, T, r, sigma)) - 1

def gamma(S, K, T, r, sigma):
    return norm.pdf(d1(S, K, T, r, sigma)) / (S * sigma * np.sqrt(T))

def theta_call(S, K, T, r, sigma):
    term1 = - (S * norm.pdf(d1(S, K, T, r, sigma)) * sigma) / (2 * np.sqrt(T))
    term2 = - r * K * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma))
    return term1 + term2

def theta_put(S, K, T, r, sigma):
    term1 = - (S * norm.pdf(d1(S, K, T, r, sigma)) * sigma) / (2 * np.sqrt(T))
    term2 = r * K * np.exp(-r * T) * norm.cdf(-d2(S, K, T, r, sigma))
    return term1 + term2

def vega(S, K, T, r, sigma):
    return S * np.sqrt(T) * norm.pdf(d1(S, K, T, r, sigma))

def rho_call(S, K, T, r, sigma):
    return K * T * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma))

def rho_put(S, K, T, r, sigma):
    return -K * T * np.exp(-r * T) * norm.cdf(-d2(S, K, T, r, sigma))
```

```python
def greeks_calculator(S, K, T, r, sigma, option_type="call"):
    if option_type == "call":
        return {
            "Price": call_price(S, K, T, r, sigma),
            "Delta": delta_call(S, K, T, r, sigma),
            "Gamma": gamma(S, K, T, r, sigma),
            "Theta": theta_call(S, K, T, r, sigma),
            "Vega": vega(S, K, T, r, sigma),
            "Rho": rho_call(S, K, T, r, sigma)
        }
    elif option_type == "put":
        return {
            "Price": put_price(S, K, T, r, sigma),
            "Delta": delta_put(S, K, T, r, sigma),
            "Gamma": gamma(S, K, T, r, sigma),
            "Theta": theta_put(S, K, T, r, sigma),
            "Vega": vega(S, K, T, r, sigma),
            "Rho": rho_put(S, K, T, r, sigma)
        }
    else:
        raise ValueError("option_type must be 'call' or 'put'")
```

```python
S = 100
K = 100
T = 1.0
r = 0.05
sigma = 0.2
call_greeks = greeks_calculator(S, K, T, r, sigma, option_type="call")
put_greeks = greeks_calculator(S, K, T, r, sigma, option_type="put")
print("CALL OPTION GREEKS")
for k, v in call_greeks.items():
    print(f"{k}: {v:.4f}")
print("\nPUT OPTION GREEKS")
for k, v in put_greeks.items():
    print(f"{k}: {v:.4f}")
```

```python
import numpy as np
from scipy.stats import norm

def d1(S, K, T, r, sigma):
    return (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))

def d2(S, K, T, r, sigma):
    return d1(S, K, T, r, sigma) - sigma * np.sqrt(T)
```

```python
def call_price(S, K, T, r, sigma):
    return S * norm.cdf(d1(S, K, T, r, sigma)) - K * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma))

def greeks_call(S, K, T, r, sigma):
    return {
        "Delta": norm.cdf(d1(S, K, T, r, sigma)),
        "Gamma": norm.pdf(d1(S, K, T, r, sigma)) / (S * sigma * np.sqrt(T)),
        "Theta": -(S * norm.pdf(d1(S, K, T, r, sigma)) * sigma) / (2 * np.sqrt(T))
                 - r * K * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma)),
        "Vega": S * np.sqrt(T) * norm.pdf(d1(S, K, T, r, sigma)),
        "Rho": K * T * np.exp(-r * T) * norm.cdf(d2(S, K, T, r, sigma))
    }
```

```python
portfolio = [
    {"K": 100, "qty": 10},   # Long 10 ATM calls
    {"K": 110, "qty": -5},   # Short 5 OTM calls
]
```

```python
def portfolio_greeks(S, T, r, sigma, portfolio):
    total_greeks = {"Delta": 0, "Gamma": 0, "Theta": 0, "Vega": 0, "Rho": 0}
    total_value = 0

    for opt in portfolio:
        K = opt["K"]
        qty = opt["qty"]

        price = call_price(S, K, T, r, sigma)
        greeks = greeks_call(S, K, T, r, sigma)

        total_value += qty * price
        for g in total_greeks:
            total_greeks[g] += qty * greeks[g]

    return total_value, total_greeks
```

#💡Day 46(Code): **"Python Implementation- Stationarity"**🚀

## ***PART 1 — What Are We Testing?***

**We want to answer:**

- Is this time series statistically stable over time?

**Specifically:**

- Does mean change?

- Does variance change?

- Is there a unit root (random walk behavior)?

***We use formal statistical tests for this.***

## ***PART 2 — Two Important Tests***

1️⃣ **Augmented Dickey-Fuller (ADF) Test Hypotheses:**

H₀: Series has a unit root → **Non-stationary**

H₁: **Series is stationary**

**If:**

p-value < 0.05 → Reject H₀ → **Stationary**

p-value > 0.05 → Fail to reject → **Non-stationary**

##2️⃣ ***KPSS Test:***

***KPSS flips the logic.***

**Hypotheses:**

H₀: Series is stationary

H₁: Series is non-stationary

**If:**

p-value < 0.05 → Reject stationarity

p-value > 0.05 → Stationary

## ***Why Use Both?***

***ADF tests for unit root.***

***KPSS tests for stationarity.***

**Using both gives stronger confirmation.**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller, kpss

np.random.seed(42)
n = 500

# White Noise
white_noise = np.random.normal(0, 1, n)

# Random Walk
random_walk = np.cumsum(np.random.normal(0, 1, n))

# Differenced Random Walk
diff_random_walk = np.diff(random_walk)
```

```python
def stationarity_tests(series, name):
    print(f"\nResults for {name}")
    print("-" * 40)

    # ADF Test
    adf_result = adfuller(series)
    print("ADF Statistic:", adf_result[0])
    print("ADF p-value:", adf_result[1])

    # KPSS Test
    kpss_result = kpss(series, regression='c', nlags="auto")
    print("KPSS Statistic:", kpss_result[0])
    print("KPSS p-value:", kpss_result[1])
```

```python
stationarity_tests(white_noise, "White Noise")
stationarity_tests(random_walk, "Random Walk")
stationarity_tests(diff_random_walk, "Differenced Random Walk")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate Synthetic Data
n = 100
X = 2 * np.random.rand(n, 1)
true_intercept = 3
true_slope = 2
noise = np.random.randn(n, 1)

y = true_intercept + true_slope * X + noise

# Add bias column (for theta_0)
X_b = np.c_[np.ones((n, 1)), X]

# Initialize Parameters
theta = np.random.randn(2, 1)  # [theta_0, theta_1]
learning_rate = 0.1
epochs = 1000

# Define Loss Function
def compute_loss(X, y, theta):
    n = len(y)
    predictions = X.dot(theta)
    loss = (1/n) * np.sum((y - predictions)**2)
    return loss

# Gradient Descent
loss_history = []
for epoch in range(epochs):
    predictions = X_b.dot(theta)
    errors = predictions - y
    gradients = (2/n) * X_b.T.dot(errors)
    theta = theta - learning_rate * gradients
    loss = compute_loss(X_b, y, theta)
    loss_history.append(loss)

# Results
print("Estimated Parameters:")
print("Intercept:", theta[0][0])
print("Slope:", theta[1][0])

# Visualization
plt.figure(figsize=(12,5))

# Plot regression fit
plt.subplot(1,2,1)
plt.scatter(X, y)
plt.plot(X, X_b.dot(theta), color='red')
plt.title("Linear Regression Fit")

# Plot loss convergence
plt.subplot(1,2,2)
plt.plot(loss_history)
plt.title("Loss Convergence")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.tight_layout()
plt.show()
```

* ***"Regularization (L1, L2, elastic net)"***

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate Multivariate Data
n = 200
p = 10
X = np.random.randn(n, p)
true_theta = np.array([[3], [1.5], [0], [0], [2], [0], [0], [0.5], [0], [0]])
y = X.dot(true_theta) + np.random.randn(n, 1)

# Add bias column
X_b = np.c_[np.ones((n, 1)), X]

# Initialize parameters
theta = np.random.randn(p+1, 1)
learning_rate = 0.01
epochs = 1000
lambda_ridge = 1
lambda_lasso = 0.1

# Ridge Regression (L2)
theta_ridge = theta.copy()
loss_ridge = []
for _ in range(epochs):
    predictions = X_b.dot(theta_ridge)
    errors = predictions - y
    gradients = (2/n) * X_b.T.dot(errors)

    # Do not regularize bias term
    gradients[1:] += 2 * lambda_ridge * theta_ridge[1:]
    theta_ridge -= learning_rate * gradients
    loss = np.mean(errors**2)
    loss_ridge.append(loss)

# Lasso Regression (L1)
theta_lasso = theta.copy()
loss_lasso = []
for _ in range(epochs):
    predictions = X_b.dot(theta_lasso)
    errors = predictions - y
    gradients = (2/n) * X_b.T.dot(errors)

    # Subgradient for L1
    gradients[1:] += lambda_lasso * np.sign(theta_lasso[1:])
    theta_lasso -= learning_rate * gradients
    loss = np.mean(errors**2)
    loss_lasso.append(loss)

# Elastic Net
theta_enet = theta.copy()
lambda1 = 0.1
lambda2 = 1
loss_enet = []

for _ in range(epochs):
    predictions = X_b.dot(theta_enet)
    errors = predictions - y
    gradients = (2/n) * X_b.T.dot(errors)
    gradients[1:] += lambda1 * np.sign(theta_enet[1:])
    gradients[1:] += 2 * lambda2 * theta_enet[1:]
    theta_enet -= learning_rate * gradients
    loss = np.mean(errors**2)
    loss_enet.append(loss)

# Compare Coefficients
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(theta_ridge[1:], label="Ridge")
plt.plot(theta_lasso[1:], label="Lasso")
plt.plot(theta_enet[1:], label="Elastic Net")
plt.legend()
plt.title("Coefficient Comparison")

plt.subplot(1,2,2)
plt.plot(loss_ridge, label="Ridge")
plt.plot(loss_lasso, label="Lasso")
plt.plot(loss_enet, label="Elastic Net")
plt.legend()
plt.title("Loss Convergence")

plt.tight_layout()
plt.show()

print("True Coefficients:")
print(true_theta.flatten())

print("\nRidge Coefficients:")
print(theta_ridge[1:].flatten())

print("\nLasso Coefficients:")
print(theta_lasso[1:].flatten())

print("\nElastic Net Coefficients:")
print(theta_enet[1:].flatten())
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler

# 1. Download Market Data
assets = ["RS", "TCS.NS", "WIT", "TATACOMM.NS", "SBIN.NS"]
prices = yf.download(assets, start="2020-01-01", end="2026-01-01")["Close"]
returns = prices.pct_change().dropna()

# 2. Feature Engineering
features = pd.DataFrame(index=returns.index)
features["momentum"] = returns.rolling(20).mean().mean(axis=1)
features["volatility"] = returns.rolling(20).std().mean(axis=1)
features["lag1"] = returns.mean(axis=1).shift(1)
features["lag2"] = returns.mean(axis=1).shift(2)
features = features.dropna()
# Target: next-day returns for each asset
targets = returns.shift(-1).loc[features.index]

# 3. Train/Test Split
split = int(len(features)*0.8)

X_train = features.iloc[:split]
X_test = features.iloc[split:]

Y_train = targets.iloc[:split]
Y_test = targets.iloc[split:]

# 4. Feature Scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 5. Train ML Model for Each Asset
models = {}
predictions = pd.DataFrame(index=X_test.index, columns=assets)
for asset in assets:
    model = RandomForestRegressor(n_estimators=200)
    model.fit(X_train_scaled, Y_train[asset])
    preds = model.predict(X_test_scaled)
    predictions[asset] = preds
    models[asset] = model

# 6. Convert Predictions into Portfolio Weights
# Positive predictions receive higher weights
weights = predictions.div(predictions.abs().sum(axis=1), axis=0)
# Replace NaN values
weights = weights.fillna(0)

# 7. Portfolio Returns
portfolio_returns = (weights * Y_test).sum(axis=1)
market_returns = Y_test.mean(axis=1)

# 8. Cumulative Performance
portfolio_curve = (1 + portfolio_returns).cumprod()
market_curve = (1 + market_returns).cumprod()

# 9. Plot Performance
plt.figure(figsize=(10,6))
plt.plot(portfolio_curve, label="ML Portfolio")
plt.plot(market_curve, label="Equal Weight Market")
plt.title("ML Portfolio vs Market")
plt.legend()
plt.show()

# 10. Print Portfolio Statistics
print("Average Portfolio Return:", portfolio_returns.mean())
print("Portfolio Volatility:", portfolio_returns.std())
print("Sharpe Ratio:", portfolio_returns.mean() / portfolio_returns.std())
```

## **🚀Day 77** -  ***"ML-Driven Portfolio Optimization with Risk Constraints"***

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Download Data for a Diverse Universe
assets = ["RS", "TCS.NS", "WIT", "TATACOMM.NS", "SBIN.NS"]
data = yf.download(assets, start="2020-01-01", end="2026-01-01")['Close']

# Flatten columns if MultiIndex exists
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# 2. Feature Engineering: Risk and Momentum
returns = data.pct_change().dropna()
# Features: Annualized Return and Annualized Volatility
stats = pd.DataFrame({
    'Return': returns.mean() * 252,
    'Volatility': returns.std() * np.sqrt(252)})

# 3. Clustering Assets by Risk-Reward Profile
scaler = StandardScaler()
scaled_stats = scaler.fit_transform(stats)

kmeans = KMeans(n_clusters=3, n_init=10)
stats['Cluster'] = kmeans.fit_predict(scaled_stats)

# 4. Portfolio Optimization (Risk-Constrained)
# We allocate more weight to lower-volatility clusters
cluster_vol = stats.groupby('Cluster')['Volatility'].mean()
inv_vol = 1 / cluster_vol
cluster_weights = inv_vol / inv_vol.sum()

# Map cluster weights back to individual assets
stats['Weight'] = stats['Cluster'].map(cluster_weights) / stats.groupby('Cluster')['Cluster'].transform('count')

# 5. Visualize the Optimized Portfolio
plt.figure(figsize=(10, 6))
plt.bar(stats.index, stats['Weight'], color='teal')
plt.title('ML-Optimized Portfolio Weights (Risk-Balanced)')
plt.ylabel('Allocation Weight')
plt.show()

# 6. Resulting Portfolio Stats
port_return = np.sum(stats['Return'] * stats['Weight'])
port_vol = np.sqrt(np.dot(stats['Weight'].T, np.dot(returns.cov() * 252, stats['Weight'])))
print(f"Optimized Portfolio Return: {port_return:.2%}")
print(f"Optimized Portfolio Volatility: {port_vol:.2%}")
print(f"Sharpe Ratio: {port_return / port_vol:.2f}")
```

## **🚀Day 81** -  ***"Attention mechanisms and Transformers"***

```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, LayerNormalization, MultiHeadAttention, Dropout
from tensorflow.keras.models import Model

# 1. Load Data
ticker = "^NSEI"
data = yf.download(ticker, start="2016-01-01", end="2026-01-01")
prices = data["Close"]

# 2. Preprocess Data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(prices.values.reshape(-1,1))

# 3. Create Sequences
def create_sequences(data, window_size=60):
    X, y = [], []

    for i in range(window_size, len(data)):
        X.append(data[i-window_size:i])
        y.append(data[i])

    return np.array(X), np.array(y)

window_size = 60
X, y = create_sequences(scaled_data, window_size)

# 4. Train-Test Split
split = int(len(X)*0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# 5. Build Transformer Block
def transformer_block(inputs, head_size, num_heads, ff_dim, dropout=0.1):

    # Multi-head attention
    x = MultiHeadAttention(num_heads=num_heads, key_dim=head_size)(inputs, inputs)

    x = Dropout(dropout)(x)
    x = LayerNormalization(epsilon=1e-6)(x + inputs)

    # Feed-forward network
    ff = Dense(ff_dim, activation="relu")(x)
    ff = Dense(inputs.shape[-1])(ff)
    ff = Dropout(dropout)(ff)
    out = LayerNormalization(epsilon=1e-6)(ff + x)

    return out

# 6. Build Model
input_layer = Input(shape=(window_size, 1))

x = transformer_block(input_layer, head_size=64, num_heads=2, ff_dim=64)
x = transformer_block(x, head_size=64, num_heads=2, ff_dim=64)

x = tf.keras.layers.GlobalAveragePooling1D()(x)
x = Dense(20, activation="relu")(x)
output = Dense(1)(x)
model = Model(inputs=input_layer, outputs=output)
model.compile(optimizer="adam", loss="mean_squared_error")

# 7. Train Model
model.fit(X_train, y_train, epochs=10, batch_size=32, verbose=1)

# 8. Predictions
predictions = model.predict(X_test)
predictions = scaler.inverse_transform(predictions)
y_test_actual = scaler.inverse_transform(y_test)

# 9. Plot Results
plt.figure(figsize=(12,6))
plt.plot(y_test_actual, label="Actual Price")
plt.plot(predictions, label="Transformer Prediction")
plt.title("Transformer Time Series Prediction")
plt.legend()
plt.show()
```

## **🚀Day 95** -  ***"Derivation Of Black-Scholes Equation"***

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
S_max = 200
K = 100
T = 1.0
r = 0.05
sigma = 0.2

M = 100  # price steps
N = 100  # time steps

dS = S_max / M
dt = T / N

# Grid Initialization
S = np.linspace(0, S_max, M)
V = np.maximum(S - K, 0)  # payoff for call option

# Backward Iteration (Explicit Scheme)
for t in range(N):
    V_new = V.copy()

    for i in range(1, M-1):
        delta = (V[i+1] - V[i-1]) / (2*dS)
        gamma = (V[i+1] - 2*V[i] + V[i-1]) / (dS**2)

        V_new[i] = V[i] + dt * (
            0.5 * sigma**2 * S[i]**2 * gamma +
            r * S[i] * delta -
            r * V[i])
        V = V_new

# Plot Option Price
plt.plot(S, V)
plt.title("Option Price via Black-Scholes PDE")
plt.xlabel("Stock Price")
plt.ylabel("Option Price")
plt.show()
```

## **🚀Day 96** -  ***"Option Pricing under Black-Scholes and Jump-Diffusion"***

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
S0 = 100
K = 100
T = 1.0
r = 0.05
sigma = 0.2

# Jump parameters
lam = 0.5       # jump intensity
mu_j = -0.2     # jump mean
sigma_j = 0.1   # jump volatility

N = 1000
dt = T / N
n_sim = 10000

# 1. Black-Scholes Closed Form
def black_scholes_call(S, K, T, r, sigma):
    d1 = (np.log(S/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)

    return S * norm.cdf(d1) - K * np.exp(-r*T) * norm.cdf(d2)

bs_price = black_scholes_call(S0, K, T, r, sigma)

# 2. Jump-Diffusion Simulation
payoffs = []
for _ in range(n_sim):
    S = S0

    for t in range(N):

        # Brownian motion
        dW = np.sqrt(dt) * np.random.randn()

        # Jump component
        jump = np.random.poisson(lam * dt)

        if jump > 0:
            J = np.exp(mu_j + sigma_j * np.random.randn())
        else:
            J = 1

        # Update price
        S = S * np.exp((r - 0.5*sigma**2)*dt + sigma*dW) * J

    payoffs.append(max(S - K, 0))

jd_price = np.exp(-r*T) * np.mean(payoffs)

# Results
print("Black-Scholes Price:", bs_price)
print("Jump-Diffusion Price:", jd_price)
```

## **🚀Day 116** -  ***"Build Implied Volatility Calculator + Surface"***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.optimize import brentq

# Black-Scholes Function
def bs_call_price(S, K, T, r, sigma):
    d1 = (np.log(S/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    return S * norm.cdf(d1) - K * np.exp(-r*T) * norm.cdf(d2)

# Implied Volatility Solver
def implied_volatility(C_market, S, K, T, r):

    def objective(sigma):
        return bs_call_price(S, K, T, r, sigma) - C_market

    return brentq(objective, 1e-6, 5)

# Example Data (Synthetic Smile)
S0 = 100
r = 0.02
T = 0.5
strikes = np.linspace(70, 130, 20)

# Simulate "market prices" with smile
true_vols = 0.2 + 0.0008*(strikes - 100)**2
market_prices = [bs_call_price(S0, K, T, r, vol) for K, vol in zip(strikes, true_vols)]

# Compute Implied Volatility
implied_vols = [implied_volatility(C, S0, K, T, r) for C, K in zip(market_prices, strikes)]

# Plot Smile
plt.figure()
plt.plot(strikes, implied_vols)
plt.title("Implied Volatility Smile")
plt.xlabel("Strike")
plt.ylabel("Implied Volatility")
plt.show()
```

## **🚀Day 117** -  ***"Monte Carlo Greeks (Delta, Vega, Gamma)"***

```python
import numpy as np

# Parameters
S0 = 100
K = 100
T = 1.0
r = 0.02
sigma = 0.2
N = 100000
h = 1e-2

# Monte Carlo Function
def mc_price(S, sigma, Z):
    ST = S * np.exp((r - 0.5*sigma**2)*T + sigma*np.sqrt(T)*Z)
    payoff = np.maximum(ST - K, 0)
    return np.exp(-r*T) * np.mean(payoff)

# Common Random Numbers
Z = np.random.randn(N)

# Base price
V = mc_price(S0, sigma, Z)

# Delta
V_up = mc_price(S0 + h, sigma, Z)
V_down = mc_price(S0 - h, sigma, Z)
delta = (V_up - V_down) / (2*h)

# Gamma
gamma = (V_up - 2*V + V_down) / (h**2)

# Vega
V_up_sigma = mc_price(S0, sigma + h, Z)
V_down_sigma = mc_price(S0, sigma - h, Z)
vega = (V_up_sigma - V_down_sigma) / (2*h)

# Results
print(f"Price: {V:.4f}")
print(f"Delta: {delta:.4f}")
print(f"Gamma: {gamma:.4f}")
print(f"Vega: {vega:.4f}")
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
from arch import arch_model
from scipy.stats import norm
import matplotlib.pyplot as plt

# 1. Load Data
data = yf.download("AAPL", start="2016-01-01", end="2026-01-01")

# Flatten MultiIndex columns if they exist and then select 'Close'
data.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col for col in data.columns.values]
data = data[['Close_AAPL']].copy() # Changed to 'Close_AAPL'
data.rename(columns={'Close_AAPL': 'Close'}, inplace=True) # Rename for consistency

# Compute log returns
data["returns"] = np.log(data["Close"] / data["Close"].shift(1))
data.dropna(inplace=True)

# 2. Signal (Momentum)
window = 5
data["signal"] = np.sign(data["returns"].rolling(window).mean())
data["signal"].fillna(0, inplace=True)

# 3. GARCH Volatility (Proper Handling)

# Scale returns for stability
returns_scaled = data["returns"] * 100

model = arch_model(returns_scaled, p=1, q=1)
res = model.fit(disp="off")

# Convert back to normal scale
data["vol"] = res.conditional_volatility / 100

# Avoid division issues
data["vol"].replace(0, np.nan, inplace=True)
data["vol"].fillna(method="bfill", inplace=True)

# 4. Position Sizing (Stable Version)
data["position"] = data["signal"] / data["vol"]

# Clip extreme leverage (IMPORTANT FIX)
data["position"] = data["position"].clip(-5, 5)

# Normalize
rolling_max = data["position"].abs().rolling(20).max()
data["position"] = data["position"] / rolling_max
data["position"].fillna(0, inplace=True)

# 5. Black-Scholes Option Pricing (Robust)
def bs_call(S, K, T, r, sigma):
    if sigma <= 0 or T <= 0:
        return 0
    d1 = (np.log(S/K) + (r + 0.5*sigma**2)*T)/(sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    return S*norm.cdf(d1) - K*np.exp(-r*T)*norm.cdf(d2)

r = 0.02
T = 30/252

option_prices = []

for i in range(len(data)):
    S = data["Close"].iloc[i]
    vol = data["vol"].iloc[i]

    if np.isnan(vol):
        option_prices.append(np.nan)
    else:
        option_prices.append(bs_call(S, S, T, r, vol))

data["option_price"] = option_prices
data["option_price"].fillna(method="ffill", inplace=True)

# Option returns (safe)
data["option_returns"] = data["option_price"].pct_change().fillna(0)

# 6. Combined Strategy
theta = 0.2  # option allocation

data["strategy_returns"] = (
    data["position"].shift(1).fillna(0) * data["returns"] +
    theta * data["option_returns"].shift(1).fillna(0))

# 7. Performance Metrics
data["cum_returns"] = (1 + data["strategy_returns"]).cumprod()
sharpe = np.sqrt(252) * data["strategy_returns"].mean() / data["strategy_returns"].std()
print(f"Sharpe Ratio: {sharpe:.2f}")

# 8. Plot
plt.figure()
plt.plot(data.index, data["cum_returns"])
plt.title("Final Strategy Performance (Volatility-Aware + Option Overlay)")
plt.xlabel("Date")
plt.ylabel("Cumulative Returns")
plt.grid()
plt.show()
```
